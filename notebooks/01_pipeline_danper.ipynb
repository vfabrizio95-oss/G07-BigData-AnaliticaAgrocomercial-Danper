```python
# Verificar períodos y valores del primer cultivo

print("CAMPAÑA AGRÍCOLA:")
print(df_muestra.iloc[7, 0])

print("\nMESES Y VALORES DEL PRIMER CULTIVO:")

for col in range(3, df_muestra.shape[1]):
    mes = df_muestra.iloc[10, col]
    valor = df_muestra.iloc[16, col]

    print(f"Columna {col}: {mes} → Producción: {valor}")
```

```python
# Comparar producción total con los períodos mensuales

import pandas as pd

fila_produccion = df_muestra.iloc[16]

total_ejecutado = pd.to_numeric(
    fila_produccion.iloc[3],
    errors="coerce"
)

# Agosto a diciembre (primer bloque)
bloque_1 = pd.to_numeric(
    fila_produccion.iloc[4:9],
    errors="coerce"
)

# Enero a diciembre (segundo bloque)
bloque_2 = pd.to_numeric(
    fila_produccion.iloc[9:21],
    errors="coerce"
)

print("Total ejecutado:", total_ejecutado)
print("Suma agosto-diciembre:", bloque_1.sum())
print("Suma enero-diciembre:", bloque_2.sum())
print("Suma de los 17 meses:",
      bloque_1.sum() + bloque_2.sum())
```

```python
# Verificar los totales de producción de todos los cultivos

resultados = []

for i in range(11, len(df_muestra)):

    variable = str(df_muestra.iloc[i, 2]).strip().lower()

    if variable.startswith("produccion") or variable.startswith("producción"):

        total = pd.to_numeric(
            df_muestra.iloc[i, 3],
            errors="coerce"
        )

        meses = pd.to_numeric(
            df_muestra.iloc[i, 9:21],
            errors="coerce"
        )

        suma_meses = meses.sum(min_count=1)

        diferencia = (
            total - suma_meses
            if pd.notna(total) and pd.notna(suma_meses)
            else float("nan")
        )

        resultados.append({
            "fila_excel": i + 1,
            "total_ejecutado": total,
            "suma_12_meses": suma_meses,
            "diferencia": diferencia
        })

df_verificacion = pd.DataFrame(resultados)

print("Registros de producción encontrados:", len(df_verificacion))

display(df_verificacion.head(15))

print("\nRegistros con diferencias mayores a 0.01:")

diferencias = df_verificacion[
    df_verificacion["diferencia"].abs() > 0.01
]

print(len(diferencias))
display(diferencias.head(10))
```

```python
# Validación de producción en todos los Excel provinciales

resultados_globales = []

for archivo in archivos:
    excel = pd.ExcelFile(archivo)

    for hoja in excel.sheet_names:
        df = pd.read_excel(
            archivo,
            sheet_name=hoja,
            header=None
        )

        for i in range(11, len(df)):

            variable = str(df.iloc[i, 2]).strip().lower()

            if variable.startswith(("produccion", "producción")):

                total = pd.to_numeric(
                    df.iloc[i, 3],
                    errors="coerce"
                )

                meses = pd.to_numeric(
                    df.iloc[i, 9:21],
                    errors="coerce"
                )

                suma = meses.sum(min_count=1)

                diferencia = (
                    abs(total - suma)
                    if pd.notna(total) and pd.notna(suma)
                    else None
                )

                resultados_globales.append({
                    "archivo": archivo.name,
                    "hoja": hoja,
                    "fila_excel": i + 1,
                    "total": total,
                    "suma_meses": suma,
                    "diferencia": diferencia
                })

df_global = pd.DataFrame(resultados_globales)

print("Total de registros evaluados:", len(df_global))

print(
    "Registros con diferencias mayores a 0.01:",
    (df_global["diferencia"] > 0.01).sum()
)

print(
    "Registros sin comparación válida:",
    df_global["diferencia"].isna().sum()
)

display(
    df_global.groupby("archivo").agg(
        registros=("fila_excel", "count"),
        diferencias=("diferencia", lambda x: (x > 0.01).sum()),
        sin_comparacion=("diferencia", lambda x: x.isna().sum())
    ).reset_index()
)
```

```python
# Paso 6.9 - Clasificar hojas distritales y provinciales

clasificacion = []

for archivo in archivos:
    excel = pd.ExcelFile(archivo)

    for hoja in excel.sheet_names:

        tipo = (
            "RESUMEN PROVINCIAL"
            if hoja.strip().lower().startswith("provincia")
            else "DISTRITO"
        )

        clasificacion.append({
            "archivo": archivo.name,
            "hoja": hoja,
            "tipo": tipo
        })

df_hojas = pd.DataFrame(clasificacion)

print("Total de hojas:", len(df_hojas))

print("\nCantidad por tipo:")
print(df_hojas["tipo"].value_counts())

display(df_hojas)
```

```python
# Paso 6.10 - Extraer registros agrícolas de las hojas distritales

import pandas as pd

registros = []

for archivo in archivos:

    # Obtener provincia desde el nombre del archivo
    provincia = (
        archivo.stem
        .replace("PROVINCIA ", "")
        .replace("_2024-2025", "")
        .replace("_", " ")
    )

    excel = pd.ExcelFile(archivo)

    for hoja in excel.sheet_names:

        # Evitar incluir resúmenes provinciales
        if hoja.strip().lower().startswith("provincia"):
            continue

        df = pd.read_excel(
            archivo,
            sheet_name=hoja,
            header=None
        )

        codigo_actual = None
        cultivo_actual = None

        for i in range(11, len(df)):

            codigo = df.iloc[i, 0]
            cultivo = df.iloc[i, 1]
            variable = df.iloc[i, 2]

            # Identificar nuevo cultivo
            if pd.notna(codigo) and pd.notna(cultivo):
                codigo_actual = str(codigo).strip()
                cultivo_actual = str(cultivo).strip()

            if pd.isna(variable) or cultivo_actual is None:
                continue

            nombre_variable = str(variable).strip()

            # Extraer los valores de las 17 columnas mensuales
            for columna in range(4, 21):

                mes = df.iloc[10, columna]
                valor = pd.to_numeric(
                    df.iloc[i, columna],
                    errors="coerce"
                )

                if pd.notna(valor):

                    registros.append({
                        "provincia": provincia,
                        "distrito": hoja,
                        "codigo_cultivo": codigo_actual,
                        "cultivo": cultivo_actual,
                        "variable": nombre_variable,
                        "columna_origen": columna,
                        "mes_origen": str(mes).strip(),
                        "valor": float(valor),
                        "archivo_origen": archivo.name
                    })

df_agricola = pd.DataFrame(registros)

print("Total de registros extraídos:", len(df_agricola))
print("Columnas:", df_agricola.columns.tolist())

display(df_agricola.head(20))
```

```python
# Paso 6.11 - Validar los registros extraídos

print("TOTAL DE REGISTROS:", len(df_agricola))

print("\nREGISTROS POR PROVINCIA:")
display(
    df_agricola.groupby("provincia")
    .size()
    .reset_index(name="registros")
)

print("\nVARIABLES AGRÍCOLAS IDENTIFICADAS:")
display(
    df_agricola["variable"]
    .value_counts()
    .reset_index()
)

print("\nCULTIVOS DIFERENTES:")
print(df_agricola["cultivo"].nunique())

print("\nVALORES NEGATIVOS:")
print((df_agricola["valor"] < 0).sum())

print("\nREGISTROS CON CAMPOS CRÍTICOS VACÍOS:")
print(
    df_agricola[
        ["provincia", "distrito", "cultivo",
         "variable", "valor"]
    ].isna().sum()
)

print("\nCÓDIGOS ASOCIADOS A MÁS DE UN CULTIVO:")
codigos = (
    df_agricola.groupby("codigo_cultivo")["cultivo"]
    .nunique()
)
print((codigos > 1).sum())
```

```python
# Verificar los cultivos principales del proyecto Danper

cultivos_principales = df_agricola[
    df_agricola["cultivo"].str.contains(
        "ESPARR|ALCACHOF",
        case=False,
        na=False
    )
]

print("REGISTROS DE ESPÁRRAGO Y ALCACHOFA:")
print(len(cultivos_principales))

print("\nCULTIVOS ENCONTRADOS:")
print(cultivos_principales["cultivo"].value_counts())

print("\nDISTRIBUCIÓN POR PROVINCIA:")
display(
    cultivos_principales.groupby(
        ["provincia", "cultivo"]
    ).size().reset_index(name="registros")
)

print("\nPRIMEROS REGISTROS:")
display(cultivos_principales.head(15))
```

```python
# Paso 6.13 - Investigar registros sin código de cultivo

sin_codigo = df_agricola[
    df_agricola["codigo_cultivo"].isna()
].copy()

print("Total de registros sin código:", len(sin_codigo))

print("\nDistribución por archivo, hoja y cultivo:")
display(
    sin_codigo.groupby(
        ["archivo_origen", "distrito", "cultivo"],
        dropna=False
    ).size().reset_index(name="registros")
)

print("\nPrimeros registros encontrados:")
display(sin_codigo.head(15))
```

```python
# Paso 6.14 - Detectar códigos vacíos y espacios

codigos_vacios = df_agricola[
    df_agricola["codigo_cultivo"].isna()
    | (
        df_agricola["codigo_cultivo"]
        .astype(str)
        .str.strip()
        == ""
    )
].copy()

print("Total de registros sin código:", len(codigos_vacios))

print("\nDistribución por cultivo:")
display(
    codigos_vacios.groupby(
        ["provincia", "distrito", "cultivo"]
    ).size().reset_index(name="registros")
)

print("\nPrimeros registros:")
display(codigos_vacios.head(15))
```

```python
# Paso 6.15 - Identificar códigos sospechosos

print("Tipos de datos:")
print(df_agricola["codigo_cultivo"].map(type).value_counts())

print("\nCódigos más frecuentes:")
display(
    df_agricola["codigo_cultivo"]
    .astype(str)
    .value_counts()
    .head(15)
)

print("\nRegistros con códigos sospechosos:")

sospechosos = df_agricola[
    df_agricola["codigo_cultivo"]
    .astype(str)
    .str.strip()
    .str.lower()
    .isin(["none", "nan", "null", ""])
]

print("Total:", len(sospechosos))
display(sospechosos.head(15))
```

```python
# Paso 6.16 - Comprobación definitiva de códigos

codigos = df_agricola["codigo_cultivo"]

print("Total de registros:", len(df_agricola))
print("Códigos nulos:", codigos.isna().sum())
print("Códigos únicos:", codigos.nunique())

print("\nCódigos no numéricos:")
no_numericos = codigos[
    pd.to_numeric(codigos, errors="coerce").isna()
]

print("Cantidad:", len(no_numericos))
print(no_numericos.value_counts(dropna=False).head(15))

print("\nEjemplos:")
display(
    df_agricola.loc[
        no_numericos.index,
        ["provincia", "distrito", "cultivo", "codigo_cultivo"]
    ].head(15)
)
```

```python
# Paso 6.17 - Comparar encabezados temporales de los seis Excel

comparacion_periodos = []

for archivo in archivos:
    excel = pd.ExcelFile(archivo)

    for hoja in excel.sheet_names:
        if hoja.lower().startswith("provincia"):
            continue

        df = pd.read_excel(
            archivo,
            sheet_name=hoja,
            header=None
        )

        encabezados = tuple(
            str(x).strip()
            for x in df.iloc[10, 4:21]
        )

        comparacion_periodos.append({
            "archivo": archivo.name,
            "hoja": hoja,
            "meses": encabezados
        })

df_periodos = pd.DataFrame(comparacion_periodos)

print("Hojas revisadas:", len(df_periodos))
print("Patrones de encabezado distintos:",
      df_periodos["meses"].nunique())

for patron in df_periodos["meses"].unique():
    print("\nPatrón:", patron)
```
