```python
from pathlib import Path
import pandas as pd

ruta = Path.home() / "proyecto_danper/datasets/produccion"

archivos = sorted(ruta.glob("*.xlsx"))

print("Total de archivos:", len(archivos))

for archivo in archivos:
    print("\nArchivo:", archivo.name)
    excel = pd.ExcelFile(archivo)
    print("Hojas disponibles:", excel.sheet_names)
```

```python
# Inspeccionar las primeras filas de una hoja real

archivo = ruta / "PROVINCIA VIRU_2024-2025.xlsx"

df_muestra = pd.read_excel(
    archivo,
    sheet_name="Viru",
    header=None
)

print("Dimensiones:", df_muestra.shape)

display(df_muestra.head(20))
```

```python
# Inspeccionar encabezados y estructura de los cultivos

pd.set_option("display.max_columns", 25)
pd.set_option("display.width", 200)

print("ENCABEZADOS Y PERÍODOS")
display(df_muestra.iloc[7:12, :])

print("\nPRIMER CULTIVO Y SUS VARIABLES")
display(df_muestra.iloc[10:19, :])

print("\nÚLTIMAS FILAS DEL ARCHIVO")
display(df_muestra.tail(10))
```

```python
# Identificar columnas y períodos exactos

print("ENCABEZADOS DE LA FILA 10")
for columna in range(df_muestra.shape[1]):
    print(
        f"Columna {columna}: "
        f"{repr(df_muestra.iloc[10, columna])}"
    )

print("\nFILAS ANTERIORES AL ENCABEZADO")
for fila in range(7, 11):
    print(f"\nFila {fila}:")
    for columna in range(df_muestra.shape[1]):
        valor = df_muestra.iloc[fila, columna]
        if pd.notna(valor):
            print(f"  Columna {columna}: {repr(valor)}")
```
